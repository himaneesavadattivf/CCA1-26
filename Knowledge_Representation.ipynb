```python
# Simple Knowledge Representation Game

# -------- KNOWLEDGE BASE --------

knowledge = {
    "player_location": "Home",
    "has_key": False,
    "treasure_found": False,

    # Semantic relationships
    "connected_locations": {
        "Home": ["Forest"],
        "Forest": ["Home", "Castle", "Key Area"],
        "Castle": ["Forest", "Tunnel"],
        "Key Area": ["Forest"],

    }
}

# -------- RULES --------

def check_rules():

    # Rule 1
    if knowledge["player_location"] == "Key Area":
        knowledge["has_key"] = True
        print("🗝️ You found the KEY!")

    # Rule 2
    if knowledge["player_location"] == "Castle":

        if knowledge["has_key"]:
            knowledge["treasure_found"] = True
            print("🎉 You opened the treasure!")

        else:
            print("🚫 The castle is locked!")
            print("You need to find the key first.")


# -------- GAME LOOP --------

while not knowledge["treasure_found"]:

    print("\n----------------------")
    print("📍 Current Location:", knowledge["player_location"])
    print("🗝️ Have Key:", knowledge["has_key"])

    current_location = knowledge["player_location"]

    print("\nYou can go to:")

    for place in knowledge["connected_locations"][current_location]:
        print("-", place)

    choice = input("\nWhere do you want to go? ")

    # Check whether move is valid
    if choice in knowledge["connected_locations"][current_location]:

        knowledge["player_location"] = choice

        # AI checks its rules
        check_rules()

    else:
        print("❌ Invalid move! You cannot go there directly.")


print("\n🏆 GAME OVER!")
print("You successfully found the treasure!")
```
